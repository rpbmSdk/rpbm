```python
import pandas as pd
import csv

```

```python
# Define the file path
file_path = './Gestion Stock V4 - Stock Complet.csv'

# Initialize an empty list to store the relevant data
data = []

# Open the CSV file and read the data
with open(file_path, mode='r', encoding='utf-8') as file:
    reader = csv.reader(file)

    # Skip the first 4 lines
    for _ in range(4):
        next(reader)
    
    # Read the header (5th line)
    header = next(reader)[5:12]
    
    # Read the relevant data from column 6 to 12
    for row in reader:
        relevant_data = row[5:12]
        data.append(relevant_data)

# Create a pandas DataFrame from the extracted data
df = pd.DataFrame(data, columns=header)

# Display the DataFrame
df
```

```python
# Cleaning of the Types
def replace_Type(old_value, new_value):
    df['TYPE'] = df['TYPE'].replace(old_value, new_value)

# Fix Typos and inconsistencies
replace_Type('PaRE BRISE', 'PARE BRISE')
replace_Type('PARE-BRISE', 'PARE BRISE')
replace_Type('PARE BRISE ', 'PARE BRISE')
replace_Type('JOINTS ', 'JOINTS')
replace_Type('jOINTS', 'JOINTS')
replace_Type('ENJOLIVEURS', 'JOINTS')
replace_Type('LUNETTES ', 'LUNETTES')
replace_Type('TOIT', 'TOIT PANORAMIQUE')
replace_Type('TOIT ', 'TOIT PANORAMIQUE')
replace_Type('TOIT PANO', 'TOIT PANORAMIQUE')
replace_Type('VITRE PAVILLON', 'TOIT PANORAMIQUE')
replace_Type('RETROVISEUR', 'RETROVISEUR EXT')
replace_Type('LEVE VITRE', 'LEVE-VITRE')
replace_Type('LEVE-VITRE ', 'LEVE-VITRE')
replace_Type('MECANISME LV', 'LEVE-VITRE')
replace_Type('MECANISME', 'LEVE-VITRE')
replace_Type('OPTIQUES', 'OPTIQUE')
replace_Type('FAISCEAU', 'OPTIQUE')
replace_Type('ANTIBROUILLARD', 'OPTIQUE')
replace_Type('FEU', 'OPTIQUE')
replace_Type('PHARES ', 'OPTIQUE')
replace_Type('PHARES', 'OPTIQUE')
replace_Type('PHARE', 'OPTIQUE')
replace_Type('','AUTRES')
replace_Type('CACHE RETRO INT','CACHE RETRO')
replace_Type('CACHE RETRO','AUTRES')
replace_Type('CAMERA','AUTRES')
replace_Type('BAIE DE PARE BRISE','AUTRES')
types = df['TYPE'].unique()
types
```

```python
AutoNickelCategories = {
    "Pare-Brise": ["PARE BRISE"],
    "Lunettes": ["LUNETTES"],
    "Glace latérale": ["GLACES LATERALES", "Custode", "Déflecteur"],
    "Joints": ["JOINTS"],
    "Consommables": [],
    "Films solaires": [],
    "Autres": [
        "TOIT PANORAMIQUE",
        "RETROVISEUR EXT",
        "RETROVISEUR INT",
        "LEVE-VITRE",
        "OPTIQUE",
        "AUTRES",
    ],
    "Services": [],
    "Accessoires": [],
}
for k,values in AutoNickelCategories.items():
    for value in values:
        if value not in types:
            print(f"{value} not found in the types")
```

```python
class Category:
    def __init__(self, name:str, id:int, parent_id:int):
        self.name = name
        self.id = id
        self.parent_id = parent_id

    def __str__(self):
        return self.__dict__
        
categoriesData = agent.search_read('product.category', [], ['name', 'id', 'parent_id'])
categories = [Category(**category) for category in categoriesData]
```

```python
for categorie in categories:
    print(categorie.__dict__)
```

```python
SaleableCategory = next(categorie for categorie in categories if categorie.name == 'Saleable')
SaleableCategory.__dict__
```

```python
def ensureCategory(name, parent_id):
    # Search for the category in categories
    category = next((c for c in categories if c.name == name), None)
    # If the category does not exist, create it
    if category is None:
        category_id = agent.create('product.category', {'name': name, 'parent_id': parent_id})
        # Update the categories 
        category = agent.search_read('product.category', [('id', '=', category_id)], ['name', 'id', 'parent_id'])[0]
        category = Category(**category)
        categories.append(category)
    return category

for k,values in AutoNickelCategories.items():
    if len(values) == 0:
        continue
    if len(values) == 1:
        # Si il n'y a qu'une seule valeur, on crée directement la catégorie enfant
        k = values[0]
        category = ensureCategory(k, SaleableCategory.id)
    else:
        # Si il y a plusieurs valeurs, on crée la catégorie parente et les catégories enfants
        category = ensureCategory(k, SaleableCategory.id)    
        for child_categories in values:
            ensureCategory(child_categories, category.id)

for categorie in categories:
    print(categorie.__dict__)
     
```

```python
# Pour chaque category dans la liste des categories on attribue l'id dans le dataFrame
df['categ_id'] = SaleableCategory.id
def assignCategory(categorie:Category):
    name = categorie.name
    if name in df['TYPE'].unique():
        df.loc[df['TYPE'] == categorie.name, 'categ_id'] = categorie.id

for categorie in categories:
    assignCategory(categorie)
df['categ_id'].unique()
```
